# Thread

```python
from threading import *

def show():
    print("This is a child thread")

t = Thread(target = show())
t.start()
print("This is the parent thread")
```

```python
class MyThread(Thread):
    def run(self):
        for i in range(3):
            print("This is a child thread")
t1 = Thread(target = MyThread().run())
t1.start()
for i in range(2):
    print("This is a parent thread")
```
